```python
df = df_2019.copy()
```

```python
print("Ülke sayısı:", df['Country or region'].nunique())
print("En yüksek skor:", df['Score'].max())
print("En düşük skor:", df['Score'].min())
```

```python
df['gdp_skor_orani'] = (df['GDP per capita'] / df['Score'] * 100).round(1)
df[['Country or region', 'Score', 'GDP per capita', 'gdp_skor_orani']].head()
```

```python
df['mutluluk_kategorisi'] = df['Score'].apply(lambda x: 'Yüksek' if x > 6 else 'Düşük')
df[['Score', 'mutluluk_kategorisi']].head()
```

```python
df_yuksek = df[df['mutluluk_kategorisi'] == 'Yüksek'].head(3)
df_dusuk = df[df['mutluluk_kategorisi'] == 'Düşük'].head(3)

df_birlesik = pd.concat([df_yuksek, df_dusuk])
df_birlesik[['Country or region', 'mutluluk_kategorisi', 'Score']]
```
